# 가격·할인율·평점이 데이터 품질에 미치는 영향 분석
> Amazon 데이터 기반 데이터 품질 관리 관점 분석

---

## 분석 목적
가격·할인율·평점 데이터의 관계를 분석하고,
AI 학습에 활용 가능한 고품질 데이터 기준을 도출하는 것이 목표.

---

## 핵심 인사이트

### 1. 과도한 할인은 데이터 신뢰도 하락으로 이어짐
할인율 81~100% 구간 평균 평점 3.93으로 가장 낮음.
Computers&Accessories(54%), Electronics(50.8%) 등
고할인율 카테고리에서 저품질 데이터 비율이 높음.

### 2. 고가 상품일수록 높은 신뢰 지표 형성
10K+ 구간 평균 평점 4.20으로 가장 높음.
가격 구간이 데이터 품질 필터링 기준으로 활용 가능.

### 3. 리뷰 데이터의 36.5%가 AI 학습에 부적합
전체 리뷰 중 36.5%가 텍스트 길이 부족 또는 URL 포함으로
AI 학습 데이터로 바로 사용 불가.
HomeImprovement(50%), OfficeProducts(45.2%) 카테고리 우선 개선 필요.

---

## 데이터 품질 개선 액션

### 1. 카테고리별 저품질 데이터 필터링 기준 설정
저품질 비율이 높은 카테고리 중심으로
최소 글자수 기준 및 URL 제거 전처리 자동화 적용.
→ AI 학습 가능 데이터 비율 63.5% → 80% 목표.

### 2. 가격·할인율 기반 데이터 품질 등급화
할인율 구간별 평점 분포를 기반으로
데이터 신뢰도 등급 기준 설계 가능.
→ RAG 서비스 응답 품질 향상에 직접 활용 가능.

---

## 가설 수립 및 검증

### 가설 1. 할인율이 높을수록 데이터 신뢰도가 낮을 것이다
- 분석: 할인율 구간별 평균 평점 측정
- 결과: 81~100% 구간 평균 평점 3.93으로 가장 낮음
- 검증: 가설 채택

### 가설 2. 가격이 높을수록 데이터 신뢰도가 높을 것이다
- 분석: 가격 구간별 평균 평점 측정
- 결과: 10K+ 구간 평균 평점 4.21로 가장 높음
- 검증: 가설 채택

### 가설 3. 리뷰 텍스트가 짧을수록 AI 학습에 부적합할 것이다
- 분석: Q1(492자) 기준 저품질 분류
- 결과: 전체의 25.2%가 길이 기준 저품질
- 검증: 가설 채택

## 한계 및 회고

실제 구매 로그 부재로 합성 데이터 기반 분석 진행.
리뷰 텍스트 품질 분석을 통해 전체의 36.5%가 AI 학습에 부적합함을 확인했으나,
감성 분석까지 진행하면 신뢰도 하락 원인을 더 구체적으로 파악 가능.
인도 루피 기준 데이터로 한국 시장 적용 시 가격 구간 재설계 필요.
