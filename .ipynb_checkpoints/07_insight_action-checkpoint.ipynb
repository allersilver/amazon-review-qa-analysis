# 할인율·평점·가격대가 구매 전환에 미치는 영향 분석
> Amazon 데이터 기반 리셀 플랫폼 관점 분석

---

## 비즈니스 문제 정의
리셀 플랫폼에서 가격은 구매 결정의 핵심 요소.
단순 할인율 상승 전략이 오히려 신뢰도 하락으로 이어질 가능성 존재.
가격/할인율/평점의 관계 파악 및 카테고리별 최적 가격 전략 도출이 목표.

---

## KPI
| 지표 | 정의 |
|------|------|
| 구매 전환율 | 조회 → 구매 비율 |
| 재구매율 | 2회 이상 구매 유저 비율 |
| 카테고리별 평균 평점 | 신뢰도 지표 |
| 할인율 구간별 전환율 | 가격 전략 효율 측정 |

---

## 핵심 인사이트

### 1. 과도한 할인은 신뢰도 하락으로 이어짐
할인율 81~100% 구간 평균 평점 3.98로 가장 낮음.
Computers&Accessories(54%), Electronics(50.8%) 등
고할인율 카테고리에서 두드러진 현상.
무조건적인 할인 경쟁보다 품질과 신뢰도 관리가 선행 필요.

### 2. 고가 상품일수록 높은 신뢰 지표 형성
10K+ 구간 평균 평점 4.20으로 가장 높음.
신중한 구매 검토로 인한 높은 만족도가 원인으로 추정.
고가 상품 과도한 할인은 오히려 의심 유발 가능성 존재.

### 3. 첫 구매 후 84% 이탈
전체 구매자의 84.1%가 1회 구매 후 재구매 없음.
재구매율 15.9%는 개선 여지가 큰 상태.
첫 구매 경험이 재구매 결정의 핵심 변수.

### 4. 리뷰 데이터의 36.5%가 AI 학습에 부적합
전체 리뷰 중 36.5%가 텍스트 길이 부족 또는 URL 포함으로 AI 학습 데이터로 바로 사용 불가.
특히 HomeImprovement(50%), OfficeProducts(45.2%) 카테고리가 저품질 비율이 높아 우선 개선 필요.

---

## 서비스 액션 제안

### 1. 카테고리별 할인율 가이드라인 설정
초기 적정 할인율 유지로 품질 및 신뢰도 우선 확보.
평점 안정화 이후 단계적 할인율 상향 조정.
→ 신뢰도 기반 장기적 구매자 유치 전략.

### 2. 고가 상품 이벤트성 쿠폰 전략
직접 할인 대신 쿠폰/적립금 방식 활용.
정가 유지를 통한 신뢰도 보호 및 구매 유인 동시 제공.
→ 고가 상품 신뢰도 유지 + 구매 전환율 개선.

### 3. 재구매 유도 알림 전략
첫 구매 후 일정 기간 내 재방문 시 쿠폰 제공.
관심 카테고리 신상품 및 가격 변동 알림 발송.
서비스 개선 사항 안내를 통한 재방문 동기 부여.
→ 재구매율 개선 → GMV 직접적 기여.

### 4. 카테고리별 리뷰 수집 가이드라인 설정
저품질 비율이 높은 카테고리 중심으로
최소 글자수 기준 및 URL 제거 전처리 자동화 적용.
→ AI 학습 가능 데이터 비율 63.5% → 80% 목표.

## 기대효과
| 액션 | 기대효과 |
|------|----------|
| 할인율 가이드라인 | 평균 평점 0.1~0.2점 개선 |

| 재구매 알림 | 재구매율 15.9% → 20% 목표 |

## 한계 및 회고

실제 구매 로그 부재로 합성 데이터 기반 퍼널/리텐션 분석 진행.  
실제 서비스 데이터였다면 이탈 구간 원인의 변수별 교차 분석 가능.  
인도 루피 기준 데이터로 한국 시장 적용 시 원화 기준 가격 구간 재설계 필요.


리뷰 텍스트 품질 분석을 통해 전체의 36.5%가 AI 학습에 부적합함을 확인했으나,  
감성 분석까지 진행하면 신뢰도 하락 원인을 더 구체적으로 파악 가능.

---

## AARRR 프레임워크 분석

서비스 성장 단계별로 데이터를 재해석했습니다.

### Acquisition (유입)
전체 분석 대상 유저 수: 9,050명
합성 데이터 기반으로 실제 유입 채널 구분은 어려우나,
실제 서비스라면 채널별(검색/광고/추천) 전환율 비교가 핵심.

### Activation (첫 구매 전환)
조회 → 장바구니 전환율: 61.3%
장바구니 → 구매 전환율: 50.2%
전체 전환율 (조회→구매): 30.8%
→ 장바구니 이탈 38.7%가 핵심 개선 구간.

### Retention (재구매)
재구매율: 15.9%
84.1%가 1회 구매 후 이탈.
→ 첫 구매 경험 개선이 리텐션의 핵심 변수.

### Revenue (매출 기여)
실제 매출 데이터 부재로 직접 측정 불가.
다만 고가 상품(10K+) 평점이 4.21로 가장 높아
프리미엄 카테고리 중심 전략이 GMV 기여 가능성 높음.

### Referral (자연 추천)
평점 4.2~4.4 구간에서 리뷰 수가 가장 많음.
→ 해당 평점 구간 유지가 자연 추천 유도에 효과적.

## 가설 수립 및 검증

### 가설 1. 할인율이 높을수록 소비자 만족도가 낮을 것이다
- 분석: 할인율 구간별 평균 평점 측정
- 결과: 81~100% 구간 평균 평점 3.93으로 가장 낮음
- 검증: ✅ 가설 채택

### 가설 2. 가격이 높을수록 소비자 신뢰도가 높을 것이다
- 분석: 가격 구간별 평균 평점 측정
- 결과: 10K+ 구간 평균 평점 4.21로 가장 높음
- 검증: ✅ 가설 채택

### 가설 3. 리뷰 텍스트가 짧을수록 AI 학습에 부적합할 것이다
- 분석: Q1(492자) 기준 저품질 분류
- 결과: 전체의 25.2%가 길이 기준 저품질
- 검증: ✅ 가설 채택
