## 전처리 - 가격

```python
df['discounted_price'] = (
    df['discounted_price']
    .str.replace('₹', '')
    .str.replace(',', '')
    .str.strip()
    .astype(float)
)

df['actual_price'] = (
    df['actual_price']
    .str.replace('₹', '')
    .str.replace(',', '')
    .str.strip()
    .astype(float)
)


df[['discounted_price', 'actual_price']].describe()
```

## 전처리 - 할인율

```python
df['discount_percentage'] = (
    df['discount_percentage']
    .str.replace('%', '')
    .str.strip()
)
df['discount_percentage'] = pd.to_numeric(df['discount_percentage'], errors='coerce')

print(df['discount_percentage'].head())
```

## 가격 구간 만들기

```python
df['discount_amount'] = df['actual_price'] - df['discounted_price']

df[['actual_price', 'discounted_price', 'discount_amount']].head()
```

```python
bins   = [0, 500, 1000, 3000, 10000, 9999999]
labels = ['~500', '500~1K', '1K~3K', '3K~10K', '10K+']

df['price_tier'] = pd.cut(df['discounted_price'], bins=bins, labels=labels)

ratio1 = (df['price_tier'].value_counts(normalize = True) * 100).round(2)
ratio1
```

전체 상품의 약 39%가 저가 상품에 집중되어 있음, 그러나 매출 기여도에서는 고가 구간 상품이 1건당 금액의 크기 때문에 더 높을 수 있음.

## 할인율 구간 만들기

```python
disc_bins   = [0, 20, 40, 60, 80, 100]
disc_labels = ['0~20%', '21~40%', '41~60%', '61~80%', '81~100%']

df['discount_tier'] = pd.cut(df['discount_percentage'], bins=disc_bins, labels=disc_labels)

ratio2 = (df['discount_tier'].value_counts(normalize=True).sort_index() * 100).round(1)
ratio2
```

- 전체 상품의 약 36%가 41%-60%에 집중되어 있음.
- 아마존은 중간 할인율 상품이 주력임을 알 수 있음.

## 평점 구간 만들기

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating'] = df['rating'].fillna(df['rating'].mean().round(1))
```

이상값 발견 '|' -> pd.to_numeric으로 처리

```python
rating_bins   = [0, 3.0, 3.5, 4.0, 4.5, 5.0]
rating_labels = ['~3.0', '3.0~3.5', '3.5~4.0', '4.0~4.5', '4.5~5.0']

df['rating_tier'] = pd.cut(df['rating'], bins=rating_bins, labels=rating_labels)

ratio3 = (df['rating_tier'].value_counts(normalize=True).sort_index() * 100).round(1)
ratio3
```

전체 상품 61.5%가 4.0-4.5에 몰려있음 -> 평점이 전반적으로 높지만 신뢰도 이슈가 있을 수 있음, 이후 평점별 구매 전환율 차이 확인 필요

## 카테고리 파싱

```python
df['category_main'] = df['category'].str.split('|').str[0]
df['category_sub']  = df['category'].str.split('|').str[1]

df[['category', 'category_main', 'category_sub']].head(3)
```

## 중복 리뷰 제거

```python
before = len(df)
df = df.drop_duplicates(subset=['review_title', 'review_content'])
after = len(df)

print(f"제거 전: {before}건 → 제거 후: {after}건 ({before - after}건 제거)")
```

```python
df.to_csv('amazon_cleaned.csv', index=False)
print("저장 완료!")
```
